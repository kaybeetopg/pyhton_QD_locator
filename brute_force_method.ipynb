```python
# Wyniki z programu ImageJ

px_kropki = [323, 519, 576, 482]
py_kropki = [271, 404, 132, 43]

x_0 = 255
y_0 = 381
x_80 = 579
y_80 = 49

s_x = 80.0/(x_80 - x_0)
s_y = 80.0/(y_0 - y_80)

print("Wyniki z programu ImageJ")

for i in range(len(px_kropki)):
    x_i = (px_kropki[i] - x_0)*s_x
    y_i = (-py_kropki[i] + y_0)*s_y
    print(f'Wspolrzedne kropki {i} w mikrometrach: ({x_i}, {y_i})')
```
